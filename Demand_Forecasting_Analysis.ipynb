# E-Commerce Demand Forecasting Using Machine Learning

## 1. Load and inspect the data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("../data/sales_data.csv")
df["Date"] = pd.to_datetime(df["Date"])
df.shape, df.head()
```

```python
print(df.dtypes)
print("\nMissing values:\n", df.isna().sum())
print("\nDate range:", df["Date"].min(), "to", df["Date"].max())
print("\nDemand summary:\n", df["Demand"].describe())
```

## 2. Time-series exploration

```python
daily = df.groupby("Date")["Demand"].mean()
plt.figure(figsize=(12,5))
plt.plot(daily.index, daily.values)
plt.title("Average Daily Demand")
plt.xlabel("Date")
plt.ylabel("Demand")
plt.show()
```

## 3. Calendar feature engineering

The first version intentionally avoids manually created lag and rolling features.

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Day"] = df["Date"].dt.day
df["Day_of_Week"] = df["Date"].dt.dayofweek
df["Week_of_Year"] = df["Date"].dt.isocalendar().week.astype(int)
df["Is_Weekend"] = (df["Day_of_Week"] >= 5).astype(int)
df["Quarter"] = df["Date"].dt.quarter
df.head()
```
